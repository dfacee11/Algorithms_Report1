## Find max value in the array

<div style="font-family: serif; font-size: 17px; line-height: 1.45; margin: 12px 0 22px 0;">
<b>ALGORITHM</b>&nbsp;&nbsp; <i>FindMax(A[0..n − 1])</i><br>
&nbsp;&nbsp;&nbsp;&nbsp;//Input: Array A[0..n − 1] of numbers<br>
&nbsp;&nbsp;&nbsp;&nbsp;//Output: Maximum value in A<br>
&nbsp;&nbsp;&nbsp;&nbsp;max ← A[0]<br>
<b>for</b> i ← 1 <b>to</b> n − 1 <b>do</b><br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<b>if</b> A[i] &gt; max<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;max ← A[i]<br>
<b>return</b> max
</div>

## Calculate an average of array numbers

<div style="font-family: serif; font-size: 17px; line-height: 1.45; margin: 12px 0 22px 0;">
<b>ALGORITHM</b>&nbsp;&nbsp; <i>Average(A[0..n − 1])</i><br>
&nbsp;&nbsp;&nbsp;&nbsp;//Input: Array A[0..n − 1] of numbers<br>
&nbsp;&nbsp;&nbsp;&nbsp;//Output: Average of the array elements<br>
&nbsp;&nbsp;&nbsp;&nbsp;sum ← 0<br>
<b>for</b> i ← 0 <b>to</b> n − 1 <b>do</b><br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;sum ← sum + A[i]<br>
&nbsp;&nbsp;&nbsp;&nbsp;average ← sum / n<br>
<b>return</b> average
</div>

## Ex. 1.1.7

Let

\[
m=qn+r,\qquad r=m\bmod n.
\]

If \(d\) divides both \(m\) and \(n\), then \(d\) divides \(m-qn=r\).

Conversely, if \(d\) divides \(n\) and \(r\), then it divides \(qn+r=m\).

Therefore both pairs have the same common divisors, and

\[
\gcd(m,n)=\gcd(n,m\bmod n).
\]

## Ex. 1.1.8

If \(m<n\), then

\[
m\bmod n=m.
\]

After the first iteration,

\[
(m,n)\rightarrow(n,m).
\]

So the algorithm effectively swaps the numbers. This can happen at most once.

## Ex. 1.2.5

### a.

Divide the decimal integer by 2 and record the remainder. Continue dividing the quotient by 2 until the quotient becomes 0. Read the remainders in reverse order to obtain the binary representation.

### b.

<div style="font-family: serif; font-size: 17px; line-height: 1.45; margin: 12px 0 22px 0;">
<b>ALGORITHM</b>&nbsp;&nbsp; <i>DecimalToBinary(n)</i><br>
&nbsp;&nbsp;&nbsp;&nbsp;//Input: Positive decimal integer n<br>
&nbsp;&nbsp;&nbsp;&nbsp;//Output: Binary representation of n<br>
&nbsp;&nbsp;&nbsp;&nbsp;B ← empty sequence<br>
<b>while</b> n &gt; 0 <b>do</b><br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;r ← n <b>mod</b> 2<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;append r to B<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;n ← ⌊n / 2⌋<br>
&nbsp;&nbsp;&nbsp;&nbsp;reverse B<br>
<b>return</b> B
</div>

&nbsp;&nbsp;&nbsp;&nbsp;<b>return</b> “Amount cannot be dispensed”<br>

&nbsp;&nbsp;&nbsp;&nbsp;d ← A[i + 1] − A[i]<br>
&nbsp;&nbsp;&nbsp;&nbsp;<b>if</b> d &lt; dmin<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;dmin ← d<br>
<b>return</b> dmin
</div>
